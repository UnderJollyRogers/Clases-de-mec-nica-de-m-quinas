# Ejemplo 2 — Derivación simbólica de las ecuaciones de velocidad

En este cuaderno se construyen las ecuaciones de cierre de un mecanismo de cuatro barras y se derivan simbólicamente con respecto al tiempo utilizando **SymPy**.

El objetivo es observar cómo aparecen las velocidades angulares

$$\dot{\theta}_2,\qquad \dot{\theta}_3,\qquad \dot{\theta}_4$$

como consecuencia de la regla de la cadena.

## 1. Variables simbólicas

Definimos el tiempo como variable independiente, las longitudes de los eslabones como constantes y los ángulos como funciones del tiempo.

```python
import sympy as sp

# Variable independiente
t = sp.symbols('t')

# Longitudes de los eslabones
L1, L2, L3, L4 = sp.symbols('L1 L2 L3 L4', positive=True)

# Ángulos como funciones del tiempo
theta_2 = sp.Function('theta_2')(t)
theta_3 = sp.Function('theta_3')(t)
theta_4 = sp.Function('theta_4')(t)

display(theta_2, theta_3, theta_4)
```

## 2. Ecuaciones de cierre

Las ecuaciones de posición del mecanismo son

$$L_2\cos\theta_2 + L_3\cos\theta_3 - L_4\cos\theta_4 - L_1 = 0,$$

$$L_2\sin\theta_2 + L_3\sin\theta_3 - L_4\sin\theta_4 = 0.$$

Las agrupamos en el vector $\mathbf{F}$.

```python
F = sp.Matrix([
    L2*sp.cos(theta_2) + L3*sp.cos(theta_3) - L4*sp.cos(theta_4) - L1,
    L2*sp.sin(theta_2) + L3*sp.sin(theta_3) - L4*sp.sin(theta_4)
])

display(F)
```

## 3. Derivar con respecto al tiempo

Ahora derivamos el vector $\mathbf{F}$ con respecto a $t$.

Como SymPy sabe que $\theta_2$, $\theta_3$ y $\theta_4$ dependen del tiempo, aplica automáticamente la regla de la cadena.

```python
Vel = sp.diff(F, t)

display(Vel)
```

Compare el resultado con las ecuaciones desarrolladas analíticamente:

$$-L_2\sin\theta_2\,\dot{\theta}_2
-L_3\sin\theta_3\,\dot{\theta}_3
+L_4\sin\theta_4\,\dot{\theta}_4=0,$$

$$L_2\cos\theta_2\,\dot{\theta}_2
+L_3\cos\theta_3\,\dot{\theta}_3
-L_4\cos\theta_4\,\dot{\theta}_4=0.$$

Identifique en cada término la derivada interna generada por la regla de la cadena.

## 4. Reemplazar las derivadas por velocidades angulares

Para hacer la notación más compacta, definimos

$$\omega_i=\dot{\theta}_i.$$

```python
omega_2, omega_3, omega_4 = sp.symbols('omega_2 omega_3 omega_4')

Vel_omega = Vel.subs({
    sp.diff(theta_2, t): omega_2,
    sp.diff(theta_3, t): omega_3,
    sp.diff(theta_4, t): omega_4
})

display(Vel_omega)
```

## 5. Forma matricial

Las incógnitas del análisis de velocidad son $\omega_3$ y $\omega_4$, mientras que $\omega_2$ es la velocidad de entrada.

SymPy puede reorganizar automáticamente las ecuaciones en la forma

$$\mathbf{J}
\begin{bmatrix}
\omega_3\\
\omega_4
\end{bmatrix}
=\mathbf{b}.$$

```python
J, b = sp.linear_eq_to_matrix(
    Vel_omega,
    [omega_3, omega_4]
)

print('Matriz Jacobiana:')
display(J)

print('Vector conocido:')
display(b)
```

## 6. Observación

La matriz obtenida es

$$\mathbf{J}=
\begin{bmatrix}
-L_3\sin\theta_3 & L_4\sin\theta_4\\
L_3\cos\theta_3 & -L_4\cos\theta_4
\end{bmatrix},$$

que corresponde a la misma matriz Jacobiana utilizada en el análisis de posiciones.
