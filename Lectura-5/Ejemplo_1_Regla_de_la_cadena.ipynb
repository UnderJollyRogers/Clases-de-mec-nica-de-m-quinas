# Ejemplo 1 — Regla de la cadena y derivación con respecto al tiempo

En este cuaderno utilizaremos **SymPy** para derivar expresiones en las que el ángulo es una función del tiempo.

El objetivo es observar cómo aparece la velocidad angular

$$\dot{\theta}=\frac{d\theta}{dt}$$

al aplicar la **regla de la cadena**.

## 1. Definir la variable independiente y la función $\theta(t)$

Primero definimos el tiempo $t$ como variable independiente y luego indicamos que $\theta$ es una función del tiempo.

```python
import sympy as sp

# Variable independiente
t = sp.symbols('t')

# theta es una función del tiempo
theta = sp.Function('theta')(t)

display(theta)
```

## 2. Derivada de $\cos\theta(t)$

Considere

$$f(t)=\cos\theta(t).$$

De acuerdo con la regla de la cadena,

$$\frac{d}{dt}\left(\cos\theta\right)
=-\sin\theta\,\frac{d\theta}{dt}.$$

Ahora obtendremos el mismo resultado simbólicamente.

```python
f = sp.cos(theta)

df_dt = sp.diff(f, t)

display(f)
display(df_dt)
```

Observe que SymPy reconoce que $\theta$ depende del tiempo y, por tanto, introduce automáticamente el término

$$\frac{d\theta}{dt}.$$

Definiendo

$$\omega=\dot{\theta}=\frac{d\theta}{dt},$$

podemos escribir el resultado en términos de la velocidad angular.

```python
omega = sp.symbols('omega')

df_dt_omega = df_dt.subs(sp.diff(theta, t), omega)

display(df_dt_omega)
```

## 3. Derivada de $\sin\theta(t)$

Modifique ahora la expresión y calcule

$$\frac{d}{dt}\left(\sin\theta(t)\right).$$

Antes de ejecutar la siguiente celda, intente anticipar el resultado.

```python
g = sp.sin(theta)

dg_dt = sp.diff(g, t)

display(g)
display(dg_dt)
```

```python
dg_dt_omega = dg_dt.subs(sp.diff(theta, t), omega)

display(dg_dt_omega)
```

## 4. Derivada de $L\cos\theta(t)$

Considere ahora un eslabón de longitud constante $L$ y la expresión

$$h(t)=L\cos\theta(t).$$

Como $L$ es constante,

$$\frac{dL}{dt}=0.$$

Calcule la derivada y observe qué términos aparecen.

```python
# L se define como una constante simbólica
L = sp.symbols('L', positive=True)

h = L*sp.cos(theta)

dh_dt = sp.diff(h, t)

display(h)
display(dh_dt)
```

```python
dh_dt_omega = dh_dt.subs(sp.diff(theta, t), omega)

display(dh_dt_omega)
```
