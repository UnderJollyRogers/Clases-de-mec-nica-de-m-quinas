# Uso de fsolve  

fsolve es una función adicional que se encuentra en la librería optimize, esta función implementa Newton raphson como se encuentra en MINPACK. En esta implementación fsolve hace una modificación a Newton raphson para evitar que el algoritmo sea divergente. En resumen:

Cerca de la raíz: converge como Newton (rápido).

Lejos de la raíz: usa mecanismos de trust region para evitar pasos “locos” (más robusto que Newton directo).

```python
import numpy as np
from scipy.optimize import fsolve

def fourbar_eqs(x, L1, L2, L3, L4, theta2):
    theta3, theta4 = x
    F1 = L2*np.cos(theta2) + L3*np.cos(theta3) - L4*np.cos(theta4) - L1
    F2 = L2*np.sin(theta2) + L3*np.sin(theta3) - L4*np.sin(theta4)
    return [F1, F2]

# Parámetros
L1, L2, L3, L4 = 100.0, 20.0, 80.0, 60.0
theta2_deg = 110
theta2 = np.deg2rad(theta2_deg)

# Semilla (en rad)
x0 = np.deg2rad([30, 90])

sol, info, ier, msg = fsolve(fourbar_eqs, x0, args=(L1, L2, L3, L4, theta2), full_output=True)

if ier != 1:
    raise RuntimeError(f"No convergió: {msg}")

theta3_sol, theta4_sol = np.rad2deg(sol)
print(f"Theta3: {theta3_sol:.4f} grados")
print(f"Theta4: {theta4_sol:.4f} grados")
```
