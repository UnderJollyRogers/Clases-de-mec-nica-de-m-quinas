# Solución del mecanismo de cuatro barras utilizando Newton Raphson

```python
import numpy as np

def newton_raphson_4bar(L1, L2, L3, L4, theta2_deg, theta_semilla, tol=1e-6, max_iter=100):
    """
    Resuelve el sistema de ecuaciones del mecanismo de cuatro barras
    utilizando el método de Newton-Raphson.
    
    Parámetros:
    - L1, L2, L3, L4: longitudes de los eslabones del mecanismo.
    - theta2_deg: ángulo de entrada (en grados).
    - tol: tolerancia para la convergencia.
    - max_iter: número máximo de iteraciones.
    - theta_semilla: valores semilla de theta3 y theta4 en forma de lista, e.g., [theta3, theta4]
    
    Retorna:
    - theta3, theta4 en radianes.
    """
    
    # Convertir ángulo de entrada a radianes
    theta2 = np.radians(theta2_deg)
    
    # Estimaciones iniciales
    theta3 = np.radians(theta_semilla[0])  # Suposición inicial
    theta4 = np.radians(theta_semilla[1])  # Suposición inicial
    
    for _ in range(max_iter):
        # Evaluar las funciones
        F1 = L2 * np.cos(theta2) + L3 * np.cos(theta3) - L4 * np.cos(theta4) - L1
        F2 = L2 * np.sin(theta2) + L3 * np.sin(theta3) - L4 * np.sin(theta4)
        
        # Construir la matriz Jacobiana
        J = np.array([
            [-L3 * np.sin(theta3), L4 * np.sin(theta4)],
            [L3 * np.cos(theta3), -L4 * np.cos(theta4)]
        ])
        
        # Resolver el sistema J * delta_theta = -F
        F = np.array([F1, F2])
        delta_theta = np.linalg.solve(J, F)
        
        # Actualizar los valores
        theta3 = theta3 - delta_theta[0]
        theta4 = theta4 - delta_theta[1]
        
        # Verificar convergencia
        if np.linalg.norm(delta_theta) < tol:
            break
    
    return np.degrees(theta3), np.degrees(theta4)  # Convertir a grados

# Parámetros del mecanismo
L1 = 100.0  # Longitud del eslabón fijo
L2 = 20.0  # Eslabón de entrada
L3 = 80.0  # Eslabón de acoplamiento
L4 = 60.0  # Eslabón de salida

# Ángulo de entrada
theta2_deg = 110  # en grados

# Valor semilla
theta_semilla = [30, 90]

# Resolver el sistema
theta3_sol, theta4_sol = newton_raphson_4bar(L1, L2, L3, L4, theta2_deg, theta_semilla)

# Mostrar resultados
print(f"Theta3: {theta3_sol:.4f} grados")
print(f"Theta4: {theta4_sol:.4f} grados")
```
