# Representación vectorial de un collarín que se mueve por una guía con orientación fija

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from matplotlib.patches import Arc
from IPython.display import HTML

# ---------------------------------------------------------
# Parámetros
# ---------------------------------------------------------
alpha_deg = 35                 # Ángulo fijo de la guía
alpha = np.radians(alpha_deg)

s_min = 0.4                    # Posición mínima del collarín
s_max = 1.4                    # Posición máxima del collarín

n_frames = 160
t = np.linspace(0, 2*np.pi, n_frames)

# Movimiento oscilante del collarín
s_vals = (s_max + s_min)/2 + (s_max - s_min)/2 * np.sin(t)

# Longitud visual de la guía
L_guia = 1.7

# ---------------------------------------------------------
# Figura
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(8, 6))

ax.set_xlim(-0.2, 1.7)
ax.set_ylim(-0.1, 1.3)
ax.set_aspect('equal')
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True)

# Ejes de referencia
ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

# Origen
ax.plot(0, 0, 'ko', markersize=8)
ax.text(-0.05, -0.08, 'O', fontsize=14)

# ---------------------------------------------------------
# Guía fija
# ---------------------------------------------------------
xg = np.array([0, L_guia*np.cos(alpha)])
yg = np.array([0, L_guia*np.sin(alpha)])
ax.plot(xg, yg, 'k-', linewidth=6, solid_capstyle='round')
ax.plot(xg, yg, color='lightgray', linewidth=10, zorder=0)

# Línea del vector de posición
vector_line, = ax.plot([], [], linewidth=3, color='royalblue')

# Punta del vector
vector_head = ax.quiver(
    0, 0, 0, 0,
    angles='xy',
    scale_units='xy',
    scale=1,
    color='royalblue',
    width=0.006
)

# Collarín (marcador cuadrado)
slider, = ax.plot([], [], marker='s', markersize=18, color='dodgerblue')

# Punto A
pointA, = ax.plot([], [], 'ko', markersize=5)

# Etiqueta A
textA = ax.text(0, 0, 'A', fontsize=14)

# Texto dinámico
text_info = ax.text(
    0.02, 0.96, '',
    transform=ax.transAxes,
    fontsize=12,
    va='top'
)

# Etiqueta del vector
text_r = ax.text(0.7, 0.35, r'$\vec{r}_{OA}$', color='royalblue', fontsize=16)

# Etiqueta de s
text_s = ax.text(0.55, 0.72, r'$s$', color='royalblue', fontsize=16)

# Arco para ángulo alpha
radio_arco = 0.32
arc = Arc((0, 0), 2*radio_arco, 2*radio_arco,
          angle=0, theta1=0, theta2=alpha_deg, linewidth=2, color='black')
ax.add_patch(arc)

# Etiqueta alpha
ax.text(0.36*np.cos(alpha/2), 0.36*np.sin(alpha/2)-0.02, r'$\alpha$', fontsize=16)

# Flechas pequeñas sobre la guía indicando deslizamiento
ax.annotate('', xy=(1.45*np.cos(alpha), 1.45*np.sin(alpha)),
            xytext=(1.25*np.cos(alpha), 1.25*np.sin(alpha)),
            arrowprops=dict(arrowstyle='->', color='royalblue', lw=2))

ax.annotate('', xy=(1.15*np.cos(alpha), 1.15*np.sin(alpha)),
            xytext=(1.35*np.cos(alpha), 1.35*np.sin(alpha)),
            arrowprops=dict(arrowstyle='->', color='royalblue', lw=2))

ax.set_title('Collarín deslizándose sobre una guía fija')

# ---------------------------------------------------------
# Animación
# ---------------------------------------------------------
def actualizar(i):
    s = s_vals[i]

    x = s*np.cos(alpha)
    y = s*np.sin(alpha)

    # Línea del vector
    vector_line.set_data([0, x], [0, y])

    # Punta del vector
    vector_head.set_UVC(x, y)

    # Collarín y punto A
    slider.set_data([x], [y])
    pointA.set_data([x], [y])

    # Etiqueta A
    textA.set_position((x + 0.04, y + 0.04))

    # Posición de etiquetas s y r
    text_r.set_position((0.55*x, 0.55*y - 0.06))
    text_s.set_position((0.55*x, 0.55*y + 0.08))

    # Texto informativo
    text_info.set_text(
        rf'$\alpha = {alpha_deg:.1f}^\circ$  (constante)' + '\n' +
        rf'$s = {s:.2f}$  (variable)'
    )

    return vector_line, vector_head, slider, pointA, textA, text_info, text_r, text_s

anim = FuncAnimation(
    fig,
    actualizar,
    frames=n_frames,
    interval=50,
    blit=False,
    repeat=True
)

plt.close(fig)
HTML(anim.to_jshtml())
```
