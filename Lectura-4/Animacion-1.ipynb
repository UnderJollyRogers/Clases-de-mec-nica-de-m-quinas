# Representación vectorial de un eslabón con longitud constante y con pivote fijo

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# ---------------------------------------------------------
# Parámetros
# ---------------------------------------------------------
L = 1.0                       # Longitud del eslabón
theta = np.linspace(0, 2*np.pi, 120)

# ---------------------------------------------------------
# Figura
# ---------------------------------------------------------
fig, ax = plt.subplots(figsize=(7, 7))

ax.set_xlim(-1.3*L, 1.3*L)
ax.set_ylim(-1.3*L, 1.3*L)
ax.set_aspect('equal')

ax.set_xlabel('x')
ax.set_ylabel('y')
ax.grid(True)

# Ejes de referencia
ax.axhline(0, linewidth=1)
ax.axvline(0, linewidth=1)

# Pivote fijo O
ax.plot(0, 0, 'ko', markersize=8)
ax.text(-0.12, -0.13, 'O', fontsize=14)

# Eslabón OA
eslabon, = ax.plot([], [], linewidth=6, solid_capstyle='round')

# Punto A
punto_A, = ax.plot([], [], 'o', markersize=9)

# Vector r_OA
vector = ax.quiver(
    0, 0, 0, 0,
    angles='xy',
    scale_units='xy',
    scale=1,
    width=0.008
)

# Texto dinámico
texto_A = ax.text(0, 0, 'A', fontsize=14)

texto_theta = ax.text(
    0.03, 0.95, '',
    transform=ax.transAxes,
    fontsize=12
)

ax.set_title(
    r'Eslabón girando alrededor de un pivote fijo'
)

# ---------------------------------------------------------
# Función de animación
# ---------------------------------------------------------
def actualizar(i):

    th = theta[i]

    # Posición del punto A
    x = L*np.cos(th)
    y = L*np.sin(th)

    # Actualizar eslabón
    eslabon.set_data([0, x], [0, y])

    # Actualizar punto A
    punto_A.set_data([x], [y])

    # Actualizar vector de posición
    vector.set_UVC(x, y)

    # Etiqueta del punto A
    texto_A.set_position((1.06*x, 1.06*y))

    # Mostrar ángulo
    texto_theta.set_text(
        rf'$L={L:.1f}$   |   '
        rf'$\theta={np.degrees(th):.1f}^\circ$'
    )

    return eslabon, punto_A, vector, texto_A, texto_theta


# ---------------------------------------------------------
# Crear animación
# ---------------------------------------------------------
animacion = FuncAnimation(
    fig,
    actualizar,
    frames=len(theta),
    interval=50,
    blit=False,
    repeat=True
)

plt.close(fig)

# Mostrar dentro de Google Colab
HTML(animacion.to_jshtml())
```
